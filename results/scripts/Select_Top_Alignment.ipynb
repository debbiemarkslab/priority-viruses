```python
import pandas as pd
df_all_entries = pd.read_csv("/n/groups/marks/projects/viral_families/notebooks/navami/alignment_metrics10_06.csv")
import numpy as np
df_all_entries["neff_L"] = df_all_entries.neff /df_all_entries.seq_len
df_all_entries["neff_L_log"] = np.log10(df_all_entries.neff_L)
```

```python
#select the top alignment per DMS with prop90
df_all_entries['prop_90_neff_min'] = np.where(df_all_entries['neff_L_log'] > 0, 
                                     df_all_entries['prop_90'],
                                     0) 

def pick_row(group):
    if group["prop_90_neff_min"].nunique() == 1:
        return group.loc[group["neff_L_log"].idxmax()]
    else:
        return group.loc[group["prop_90_neff_min"].idxmax()]

df_top = (
    df_all_entries.groupby("ID", group_keys=False)
          .apply(pick_row)
          .reset_index(drop=True)
)
```
